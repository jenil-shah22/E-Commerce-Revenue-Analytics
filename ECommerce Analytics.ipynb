```python
#Monthly Revenue Trend

monthly_revenue = (
    order_items.merge(orders[["order_id", "order_purchase_timestamp"]],
                     on = "order_id",
                     how = "left")  
)

monthly_revenue["year_month"] = (
    monthly_revenue["order_purchase_timestamp"].dt.to_period("M")
)

monthly_revenue = (
    monthly_revenue
    .groupby("year_month")["total_item_value"]
    .sum()
    .reset_index()
)

monthly_revenue.head()
```

```python
products_en = products.merge(
    category_translation,
    on="product_category_name",
    how="left"
)

products_en.head()
```

```python
category_sales = order_items.merge(
    products_en[[
        "product_id",
        "product_category_name_english"
    ]],
    on="product_id",
    how="left"
)

category_sales.head()
```

```python
#Revenue by Category
category_revenue = (
    category_sales
    .groupby("product_category_name_english")["total_item_value"]
    .sum()
    .reset_index()
    .sort_values(
        by="total_item_value",
        ascending=False
    )
)

category_revenue.head(10)
```

```python
#Top Categories by Units Sold
category_units = (
    category_sales
    .groupby("product_category_name_english")
    .size()
    .reset_index(name="units_sold")
    .sort_values(
        by="units_sold",
        ascending=False
    )
)

category_units.head(10)
```

```python
#Average Revenue Per Unit by Category
category_metrics = (
    category_sales
    .groupby("product_category_name_english")
    .agg(
        revenue=("total_item_value", "sum"),
        units_sold=("product_id", "count")
    )
    .reset_index()
)

category_metrics["avg_revenue_per_unit"] = (
    category_metrics["revenue"]
    / category_metrics["units_sold"]
)

category_metrics.sort_values(
    by="avg_revenue_per_unit",
    ascending=False
).head(10)
```

```python
#Seller Revenue Analysis
seller_revenue = (
    order_items
    .groupby("seller_id")
    .agg(
        revenue=("total_item_value", "sum"),
        orders=("order_id", "nunique"),
        products_sold=("product_id", "count")
    )
    .reset_index()
)

seller_revenue.head()
```

```python
#Top Sellers by Revenue
seller_revenue.sort_values(
    by="revenue",
    ascending=False
).head(10)
```

```python
seller_revenue["revenue"].describe()
```

```python
top_10_revenue = (
    seller_revenue
    .sort_values("revenue", ascending=False)
    .head(10)["revenue"]
    .sum()
)

total_revenue = seller_revenue["revenue"].sum()

(top_10_revenue / total_revenue) * 100
```

```python
# State that generates the most orders
customer_orders = orders.merge(
    customers[[
        "customer_id", 
        "customer_state"
    ]],
    on = "customer_id",
    how = "left"
)

state_orders = (
    customer_orders
    .groupby("customer_state")["order_id"]
    .count()
    .reset_index(name="total_orders")
    .sort_values(
        by="total_orders",
        ascending=False
    )
)

state_orders.head(10)
```

```python
# States that generates the most revenue
order_revenue = (
    order_items
    .groupby("order_id")["total_item_value"]
    .sum()
    .reset_index()
)

state_revenue = (
    customer_orders
    .merge(order_revenue, on="order_id")
)

state_revenue_analysis = (
    state_revenue
    .groupby("customer_state")["total_item_value"]
    .sum()
    .reset_index()
    .sort_values(
        by="total_item_value",
        ascending=False
    )
)

state_revenue_analysis.head(10)
```

```python
#Revenue Per Order by State
state_metrics = (
    state_revenue
    .groupby("customer_state")
    .agg(
        revenue=("total_item_value", "sum"),
        orders=("order_id", "nunique")
    )
    .reset_index()
)

state_metrics["revenue_per_order"] = (
    state_metrics["revenue"]
    / state_metrics["orders"]
)

state_metrics.sort_values(
    by="revenue_per_order",
    ascending=False
).head(10)
```

```python
import pandas as pd
import plotly.express as px
from dash import Dash, html, dcc

# ==========================
# LOAD DATA
# ==========================

kpi_df = pd.read_csv("data/processed/kpi_metrics.csv")
monthly_revenue = pd.read_csv("data/processed/monthly_revenue.csv")
state_revenue = pd.read_csv("data/processed/state_revenue.csv")

# ==========================
# KPI VALUES
# ==========================

total_revenue = round(kpi_df["Total Revenue"][0], 2)
total_orders = int(kpi_df["Total Orders"][0])
total_customers = int(kpi_df["Total Customers"][0])
aov = round(kpi_df["Average Order Value"][0], 2)
avg_delivery = round(kpi_df["Average Delivery Time"][0], 2)

# ==========================
# KPI CARD FUNCTION
# ==========================

def create_card(title, value, subtitle):

    return html.Div(

        [

            html.H4(
                title,
                style={
                    "color": "#94A3B8",
                    "marginBottom": "10px"
                }
            ),

            html.H2(
                value,
                style={
                    "color": "#F8FAFC",
                    "marginBottom": "10px",
                    "fontWeight": "bold"
                }
            ),

            html.P(
                subtitle,
                style={
                    "color": "#64748B",
                    "fontSize": "14px"
                }
            )

        ],

        style={
            "backgroundColor": "#1E293B",
            "padding": "20px",
            "borderRadius": "16px",
            "width": "18%",
            "textAlign": "center",
            "border": "1px solid #334155",
            "boxShadow": "0px 4px 20px rgba(0,0,0,0.25)"
        }

    )

# ==========================
# KPI ROW
# ==========================

kpi_row = html.Div(

    [

        create_card(
            "Revenue",
            f"${total_revenue:,.0f}",
            "Marketplace Revenue"
        ),

        create_card(
            "Orders",
            f"{total_orders:,}",
            "Total Orders"
        ),

        create_card(
            "Customers",
            f"{total_customers:,}",
            "Unique Buyers"
        ),

        create_card(
            "AOV",
            f"${aov}",
            "Average Order Value"
        ),

        create_card(
            "Delivery",
            f"{avg_delivery} Days",
            "Average Delivery Time"
        )

    ],

    style={
        "display": "flex",
        "justifyContent": "space-between",
        "marginTop": "40px"
    }

)

# ==========================
# MONTHLY REVENUE CHART
# ==========================

fig_monthly = px.line(
    monthly_revenue,
    x="year_month",
    y="total_item_value",
    title="Monthly Revenue Trend",
    markers=True
)

fig_monthly.update_traces(
    line=dict(
        color="#10B981",
        width=4
    ),
    marker=dict(
        size=8,
        color="#10B981"
    )
)

fig_monthly.update_layout(

    paper_bgcolor="#0F172A",
    plot_bgcolor="#0F172A",

    font_color="#F8FAFC",

    title={
        "font": {
            "size": 22
        }
    },

    xaxis=dict(
        showgrid=False,
        color="#CBD5E1"
    ),

    yaxis=dict(
        showgrid=True,
        gridcolor="#334155",
        color="#CBD5E1"
    )

)

# ==========================
# STATE REVENUE CHART
# ==========================

fig_state = px.bar(
    state_revenue.head(10),
    x="customer_state",
    y="total_item_value",
    title="Top States by Revenue",
    color="total_item_value",
    color_continuous_scale="Tealgrn"
)

fig_state.update_layout(

    paper_bgcolor="#0F172A",
    plot_bgcolor="#0F172A",

    font_color="#F8FAFC",

    title={
        "font": {
            "size": 22
        }
    },

    xaxis=dict(
        showgrid=False,
        color="#CBD5E1"
    ),

    yaxis=dict(
        showgrid=True,
        gridcolor="#334155",
        color="#CBD5E1"
    ),

    coloraxis_showscale=False

)

# ==========================
# DASH APP
# ==========================

app = Dash(__name__)

app.layout = html.Div(

    style={

        "backgroundColor": "#0F172A",
        "minHeight": "100vh",
        "padding": "30px",
        "fontFamily": "Segoe UI"

    },

    children=[

        html.H1(

            "E-Commerce Revenue Intelligence Dashboard",

            style={
                "color": "#F8FAFC",
                "textAlign": "center",
                "fontWeight": "bold"
            }

        ),

        html.P(

            "Interactive dashboard for revenue monitoring, customer analytics, seller performance, product intelligence, and geographic insights.",

            style={
                "color": "#CBD5E1",
                "textAlign": "center",
                "fontSize": "18px",
                "marginBottom": "30px"
            }

        ),

        kpi_row,

        html.Br(),
        html.Br(),

        dcc.Graph(
            figure=fig_monthly
        ),

        dcc.Graph(
            figure=fig_state
        )

    ]

)

app.run(
    debug=True,
    port=8090
)
```

```python
monthly_revenue_clean = monthly_revenue_clean.reset_index(drop=True)

monthly_revenue_clean["month_index"] = range(
    len(monthly_revenue_clean)
)
```

```python
#Train Linear Regression Model
from sklearn.linear_model import LinearRegression

X = monthly_revenue_clean[["month_index"]]
y = monthly_revenue_clean["total_item_value"]

model = LinearRegression()

model.fit(X, y)

monthly_revenue_clean["predicted_revenue"] = model.predict(X)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

y_pred = model.predict(X)

print("R2 Score:", r2_score(y, y_pred))
print("MAE:", mean_absolute_error(y, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y, y_pred)))
print("RMSE:", np.sqrt(mean_squared_error(y, y_pred)))
```

```python
future_months = pd.DataFrame(
    {
        "month_index": range(
            len(monthly_revenue_clean),
            len(monthly_revenue_clean) + 6
        )
    }
)

future_months["forecast_revenue"] = (
    model.predict(future_months)
)

future_months
```
